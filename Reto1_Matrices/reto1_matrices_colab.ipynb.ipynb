## 6. Lectura desde Google Drive y cálculo con NumPy

El archivo generado en el computador se lee desde Google Drive utilizando NumPy. Se calculan nuevamente el mínimo, máximo, media y desviación estándar para comparar estos resultados con los obtenidos manualmente en Python.

```python
import numpy as np
import time

inicio_numpy = time.perf_counter()

vector_numpy = np.loadtxt(ruta, dtype=np.uint8)

minimo_numpy = np.min(vector_numpy)
maximo_numpy = np.max(vector_numpy)
media_numpy = np.mean(vector_numpy)
desviacion_numpy = np.std(vector_numpy)

fin_numpy = time.perf_counter()

tiempo_numpy = fin_numpy - inicio_numpy

print("Cantidad de valores leídos:", vector_numpy.size)
print("Valor mínimo:", minimo_numpy)
print("Valor máximo:", maximo_numpy)
print("Media:", media_numpy)
print("Desviación estándar:", desviacion_numpy)
print("Tiempo total con NumPy:", tiempo_numpy, "segundos")
```

## 7. Visualización de la matriz como imagen

El vector de 1.000.000 de elementos se reorganiza nuevamente como una matriz de 1000x1000. Cada valor entre 0 y 255 representa la intensidad de un píxel en escala de grises.

```python
import matplotlib.pyplot as plt

imagen = vector_numpy.reshape(1000, 1000)

plt.figure(figsize=(8, 8))
plt.imshow(imagen, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen generada a partir de la matriz aleatoria")
plt.axis("off")
plt.show()
```

Los resultados obtenidos mediante la implementación manual y mediante NumPy son prácticamente iguales. La pequeña diferencia observada en los últimos decimales de la desviación estándar se debe a la precisión numérica utilizada durante las operaciones de punto flotante.

En cuanto al tiempo de ejecución, en esta prueba NumPy realizó el proceso en menor tiempo que la implementación manual.
