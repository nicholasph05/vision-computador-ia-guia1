# Reto 1 - Matrices a imágenes

## Objetivo

Crear una matriz de 1000x1000 con valores aleatorios entre 0 y 255 sin utilizar NumPy ni librerías especializadas en tratamiento matricial.

Posteriormente se calcularán manualmente el mínimo, máximo, media y desviación estándar, se medirá el tiempo de ejecución, se aplanará la matriz y se guardará en un archivo de texto para luego compararlo con una implementación usando NumPy en Google Colab.

```python
import random
import time
import math
```

## 1. Creación de la matriz 1000x1000

Se crea una matriz de 1000 filas por 1000 columnas utilizando listas de Python. Cada posición contiene un valor entero aleatorio entre 0 y 255.

```python
filas = 1000
columnas = 1000

matriz = []

for i in range(filas):
    fila = []
    
    for j in range(columnas):
        valor = random.randint(0, 255)
        fila.append(valor)
    
    matriz.append(fila)

print("Matriz creada correctamente")
print("Número de filas:", len(matriz))
print("Número de columnas:", len(matriz[0]))
```

## 2. Cálculo manual de estadísticas

Se implementa una función propia para obtener el valor mínimo, máximo, media y desviación estándar de todos los elementos de la matriz.

Los cálculos se realizan recorriendo manualmente la matriz, sin utilizar NumPy ni librerías especializadas en tratamiento matricial.

```python
def calcular_estadisticas(matriz):
    minimo = 255
    maximo = 0
    suma = 0
    cantidad = 0

    # Primer recorrido:
    # mínimo, máximo, suma y cantidad de elementos
    for fila in matriz:
        for valor in fila:
            
            if valor < minimo:
                minimo = valor
            
            if valor > maximo:
                maximo = valor
            
            suma += valor
            cantidad += 1

    # Cálculo de la media
    media = suma / cantidad

    # Segundo recorrido:
    # cálculo de la suma de diferencias al cuadrado
    suma_diferencias = 0

    for fila in matriz:
        for valor in fila:
            diferencia = valor - media
            suma_diferencias += diferencia ** 2

    # Varianza poblacional
    varianza = suma_diferencias / cantidad

    # Desviación estándar
    desviacion_estandar = math.sqrt(varianza)

    return minimo, maximo, media, desviacion_estandar
```

```python
minimo, maximo, media, desviacion = calcular_estadisticas(matriz)

print("Valor mínimo:", minimo)
print("Valor máximo:", maximo)
print("Media:", media)
print("Desviación estándar:", desviacion)
```

## 3. Medición del tiempo de ejecución

Se mide el tiempo total requerido para crear la matriz de 1000x1000 y calcular manualmente el mínimo, máximo, media y desviación estándar.

```python
inicio = time.perf_counter()

# Crear nuevamente la matriz
matriz = []

for i in range(1000):
    fila = []

    for j in range(1000):
        valor = random.randint(0, 255)
        fila.append(valor)

    matriz.append(fila)

# Calcular las estadísticas
minimo, maximo, media, desviacion = calcular_estadisticas(matriz)

fin = time.perf_counter()

tiempo_total = fin - inicio

print("Valor mínimo:", minimo)
print("Valor máximo:", maximo)
print("Media:", media)
print("Desviación estándar:", desviacion)
print("Tiempo total de ejecución:", tiempo_total, "segundos")
```

## 4. Aplanamiento de la matriz

La matriz bidimensional de 1000x1000 se convierte en un vector unidimensional que contiene los 1.000.000 de valores originales.

```python
vector = []

for fila in matriz:
    for valor in fila:
        vector.append(valor)

print("Cantidad de elementos de la matriz:", 1000 * 1000)
print("Cantidad de elementos del vector:", len(vector))

```

## 5. Almacenamiento del vector en archivo de texto

El vector aplanado se guarda en un archivo de texto plano. Cada valor se almacena en una línea independiente para facilitar su posterior lectura desde Google Colab.

```python
import os

os.makedirs("datos", exist_ok=True)

ruta_archivo = "datos/vector_matriz.txt"

with open(ruta_archivo, "w") as archivo:
    for valor in vector:
        archivo.write(str(valor) + "\n")

print("Archivo guardado correctamente.")
print("Ruta:", ruta_archivo)
print("Cantidad de valores guardados:", len(vector))
```
